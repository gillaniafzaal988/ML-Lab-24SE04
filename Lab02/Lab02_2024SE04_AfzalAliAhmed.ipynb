## Task 4 · Spacing and reshaping
- **(a)** An array from 0.01 to 1.00 in steps of 0.01, reshaped to 10 × 10.
- **(b)** 20 evenly spaced points between 0 and 1.
- **(c)** Reshape Lahore's 12 monthly rainfall values (`rain[0]`) into 4 quarters × 3 months.

## Task 6 · Census 2023 (real data, PBS)
Using `units`, `pop_2023` and `pop_2017`:
- **(a)** Total population.

- **(c)** The most populous unit.
- **(d)** Units with more than 20 million people.
- **(e)** Annual growth rate 2017–2023 for every unit, in one line: `((pop_2023 / pop_2017) ** (1/6) - 1) * 100`.
- **(f)** The fastest-growing unit.

**Interpret:** in 1–2 lines, what do (e) and (f) say? Compare AJK with the rest.

```python
# TASK 6 · CENSUS 2023 (A–F)  (uses the given units, pop_2023, pop_2017)
# A: Total population
total_pop_2023 = pop_2023.sum()
print(f"T6a: Total population 2023: {total_pop_2023:,} people")

# B: Share of each unit in %
shares = pop_2023 / total_pop_2023 * 100
print("T6b: Share of total (%):", dict(zip(units.tolist(), np.round(shares, 2).tolist())))

# C: Most populous unit
most_pop_idx = np.argmax(pop_2023)
print(f"T6c: Most populous unit: {units[most_pop_idx]} ({pop_2023[most_pop_idx]:,} people)")

# D: Units with more than 20 million people
print("T6d: Units above 20 million:", units[pop_2023 > 20_000_000])

# E: Annual growth rate 2017-2023 in one line
growth = ((pop_2023 / pop_2017) ** (1/6) - 1) * 100
print("T6e: Annual growth 2017-2023 (%/year):", dict(zip(units.tolist(), np.round(growth, 2).tolist())))

# F: Fastest-growing unit
fastest_idx = np.argmax(growth)
print(f"T6f: Fastest-growing unit: {units[fastest_idx]} ({growth[fastest_idx]:.2f} %/year)")

```

**Interpretation (Task 6):** Balochistan is the fastest-growing unit (about 3.19 % per year), followed by Islamabad (2.80 %) and Sindh (2.56 %); the larger provinces all grew about 2.3–2.6 % per year. AJK grew slowest by far (about 1.15 % per year, less than half the others), so its share of Pakistan's population is shrinking.

## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).
- **(g)** Monsoon (Jul–Sep) rain of each city by slicing columns.
- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# TASK 7 · CLIMATE (A–H)  (uses the given tmax, rain, cities, months)
# A: Shape
print("T7a: Shape of tmax:", tmax.shape)

# B: Annual mean maximum temperature of each city
print("T7b: Annual mean max temp (°C):", dict(zip(cities.tolist(), np.round(tmax.mean(axis=1), 2).tolist())))

# C: Hottest month of each city
print("T7c: Hottest month:", dict(zip(cities.tolist(), months[np.argmax(tmax, axis=1)].tolist())))
# argmax returns the FIRST maximum, so check whether any city has tied hottest months
n_at_max = (tmax == tmax.max(axis=1, keepdims=True)).sum(axis=1)
print("T7c: Months sharing the max temperature:", dict(zip(cities.tolist(), n_at_max.tolist())))
# Lahore (May = Jun = 38.9 °C) and Karachi (May = Jun = 35.7 °C) tie, so argmax reports May for both.

# D: 5-city average for each month
print("T7d: 5-city monthly average (°C):", dict(zip(months.tolist(), np.round(tmax.mean(axis=0), 2).tolist())))

# E: Number of months above 35 °C in each city
print("T7e: Months above 35 °C:", dict(zip(cities.tolist(), (tmax > 35).sum(axis=1).tolist())))

# F: Lahore in °F
lahore_f = tmax[0] * 9/5 + 32
print("T7f: Lahore monthly max temps (°F):", np.round(lahore_f, 1))

# G: Monsoon (Jul-Sep) rain by slicing columns 6:9
monsoon_rain = rain[:, 6:9]
monsoon_totals = monsoon_rain.sum(axis=1)
print("T7g: Monsoon (Jul-Sep) rain per city (mm):", dict(zip(cities.tolist(), np.round(monsoon_totals, 1).tolist())))

# H: Wettest monsoon city and monsoon share of yearly rain
wettest_idx = np.argmax(monsoon_totals)
print(f"T7h: Wettest monsoon city: {cities[wettest_idx]} ({monsoon_totals[wettest_idx]:.1f} mm)")
share_pct = monsoon_totals / rain.sum(axis=1) * 100
print("T7h: Monsoon share of yearly rain (%):", dict(zip(cities.tolist(), np.round(share_pct, 1).tolist())))

```

**Interpretation (Task 7):** Surprise: Karachi is the hottest city on average (32.5 °C) and gets 73.8 % of its yearly rain in the monsoon, the highest share of all five, yet its monsoon total (234.8 mm) is far below Islamabad's 763 mm, the wettest. Quetta also surprised me: its hottest month is July, not May or June.

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
`cpi` has 2 rows (FY2024, FY2025) and 10 columns (Jul to Apr), year-on-year CPI inflation in %.
- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.
- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# TASK 8 · INFLATION (A–E)  (uses the given cpi, cpi_months)
# A: Average inflation of each year
print(f"T8a: FY2024 average inflation (Jul-Apr): {cpi[0].mean():.2f} %")
print(f"T8a: FY2025 average inflation (Jul-Apr): {cpi[1].mean():.2f} %")

# B: FY2024 months with inflation above 25 %
print("T8b: FY2024 months above 25 %:", cpi_months[cpi[0] > 25])

# C: Drop in inflation for each month (FY2024 - FY2025)
drop = cpi[0] - cpi[1]
print("T8c: Monthly drop (percentage points):", dict(zip(cpi_months.tolist(), np.round(drop, 1).tolist())))

# D: Month with the biggest drop
big_idx = np.argmax(drop)
print(f"T8d: Biggest drop: {cpi_months[big_idx]} ({drop[big_idx]:.1f} percentage points)")

# E: Label FY2025 months: Low (<5), Moderate (5-10), High (>=10)
labels = np.where(cpi[1] < 5, "Low", np.where(cpi[1] < 10, "Moderate", "High"))
print("T8e: FY2025 labels:", dict(zip(cpi_months.tolist(), labels.tolist())))

```

**Interpretation (Task 8):** Prices kept rising in both years, but far more slowly in FY2025. Average inflation fell from about 26.2 % (FY2024) to 4.9 % (FY2025), the biggest fall was in January (25.9 percentage points), and by April 2025 inflation was only 0.3 %.

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# TASK 9 · GRACE MARKS (A–D)  (uses the given marks)
# A: Add 5 grace marks, capped at 100
graced = np.clip(marks + 5, 0, 100)
print("T9a: Marks after grace (capped at 100):", graced)

# B: Pass % (>= 50) after grace
pass_pct_before = (marks >= 50).mean() * 100
pass_pct_after = (graced >= 50).mean() * 100
print(f"T9b: Pass % before grace: {pass_pct_before:.1f} %")
print(f"T9b: Pass % after grace: {pass_pct_after:.1f} %")

# C: Index of students who passed ONLY because of the grace marks
only_grace = np.where((marks < 50) & (graced >= 50))[0]
print("T9c: Student indices who passed only due to grace:", only_grace)

# D: z-score of each student after grace
z_scores = (graced - graced.mean()) / graced.std()
print("T9d: z-scores after grace:", np.round(z_scores, 2))
```

## Task 10 · An image is an array
- **(a)** Start from `np.zeros((8, 8), dtype=int)` and draw the "7" from the lecture with slicing (white = 255, grey = 128).
- **(b)** Flip it left-right.
- **(c)** Invert it (`255 - img`).
- **(d)** Crop rows 1–4 and columns 3–6.
- **(e)** Brighten by 80, capped at 255.
- **(f)** Print its shape, dtype and number of white pixels.

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`

```python
import numpy as np
import matplotlib.pyplot as plt

# TASK 10 · IMAGE IS AN ARRAY (A–F)

# A: Draw the "7" on an 8 x 8 canvas matching the lecture slide
img = np.zeros((8, 8), dtype=int)
img[1, 1:7] = 255      # Top horizontal white bar
img[2:7, 5] = 255      # Vertical white stem down column 5
img[4, 3:7] = 128      # Middle horizontal grey crossbar (value 128)
print("T10a: Canvas with 7:\n", img)

# B: Flip left-right
flipped = img[:, ::-1]  # Equivalent to np.fliplr(img)
print("T10b: Flipped Left-Right:\n", flipped)

# C: Invert canvas colors
inverted = 255 - img
print("T10c: Inverted Canvas:\n", inverted)

# D: Crop rows 1–4 and columns 3–6 (python slice 1:5, 3:7)
crop = img[1:5, 3:7]
print("T10d: Cropped Area:\n", crop)

# E: Brighten by 80, capped at 255
brightened = np.clip(img + 80, 0, 255)
print("T10e: Brightened canvas:\n", brightened)

# F: Shape, dtype, white pixels count
white_pixels_count = (img == 255).sum()
print(f"T10f: Shape: {img.shape}, Dtype: {img.dtype}, White Pixels: {white_pixels_count}")

# Optional: Visualize the 2D array as a grayscale image using Matplotlib
plt.figure(figsize=(4, 4))
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.title("Task 10: Array Visualized as Image")
plt.colorbar()
plt.show()
```
