## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# Task 3:
# (a) Missing values
print("T3a: missing values per column")
print(df.isna().sum())
rows_before = len(df)
df = df.dropna(subset=["Brand", "Model"])
df["Registered City"] = df["Registered City"].fillna("Unknown")
df["Fuel"] = df["Fuel"].fillna("Unknown")
df["Transaction Type"] = df["Transaction Type"].fillna("Unknown")
print("T3a: rows removed =", rows_before - len(df), "| remaining =", len(df))

# (b) Formatting: city names and Price text
df["Registered City"] = df["Registered City"].str.strip().str.title()
df["Price"] = (df["Price"].astype(str)
                 .str.replace("Rs", "", regex=False)
                 .str.replace(",", "", regex=False)
                 .str.strip())
print("T3b: sample cleaned prices:", df["Price"].head(3).tolist())

# (c) Data types
df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")
print("T3c: Price dtype =", df["Price"].dtype, "| Year dtype =", df["Year"].dtype)

# (d) Inconsistent categories
print("T3d: cities before =", df["Registered City"].nunique())
df["Registered City"] = df["Registered City"].replace("Sukkar", "Sukkur")
df["Model"] = df["Model"].replace("Corrolla Altis", "Corolla Altis")
print("T3d: cities after  =", df["Registered City"].nunique())

# (e) Duplicates
print("T3e: exact duplicate ads =", df.duplicated().sum())
df = df.drop_duplicates()
print("T3e: rows after dropping duplicates =", len(df))

# (f) Invalid values
print("T3f: before fixing")
print(df[["KMs Driven", "Price"]].describe())
bad_km = ((df["KMs Driven"] < 100) | (df["KMs Driven"] > 1_000_000)
          | (df["KMs Driven"] == 12345) | (df["KMs Driven"] == 123456))
df.loc[bad_km, "KMs Driven"] = np.nan
print("T3f: KMs Driven set to NaN =", bad_km.sum())

fake = (df["Price"] == 12_345_678) | (df["Price"] == 87_654_321)
cheap_brand = (df["Brand"] == "Suzuki") | (df["Brand"] == "Daihatsu")
bad_price = fake | (cheap_brand & (df["Price"] > 5_000_000))
print("T3f: ads dropped for fake prices =", bad_price.sum())
df = df[~bad_price].reset_index(drop=True)

# (g) Self-check
print("T3g: final shape =", df.shape, "(expected (20484, 9))")
```

**Note:** Year and KMs Driven were left as NaN on purpose, because we cannot know the real values. mean() and median() skip NaN.

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# Task 4:
# (a) Price in lakh rupees
df["Price_lakh"] = df["Price"] / 100_000
print("T4a:")
print(df[["Price", "Price_lakh"]].head())

# (b) Age of the car
df["Age"] = 2020 - df["Year"]
print("T4b:")
print(df[["Year", "Age"]].head())

# (c) 5 most common brands
print("T4c:")
print(df["Brand"].value_counts().head(5))

# (d) 5 cities with the most ads (Unknown is a placeholder, not a city)
known_city = df["Registered City"] != "Unknown"                                   
print("T4d:")
print(df.loc[known_city, "Registered City"].value_counts().head(5))              

# (e) Median Price_lakh by fuel type, highest first
print("T4e:")
print(df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False))

# (f) Median price of Corolla GLI for each year 2010-2018
gli = df[(df["Model"] == "Corolla GLI") & (df["Year"] >= 2010) & (df["Year"] <= 2018)]
gli_med = gli.groupby("Year")["Price_lakh"].median()
print("T4f: median Price_lakh of Corolla GLI by year")
print(gli_med)
print("T4f: average loss per year of age (lakh) =",
      round((gli_med.loc[2018] - gli_med.loc[2010]) / 8, 2))

# (g) 5 most expensive ads
print("T4g:")
print(df.nlargest(5, "Price")[["Brand", "Model", "Year", "KMs Driven", "Price_lakh"]])

# (h) Cash vs Installment/Leasing
print("T4h:")
cash_inst = df[(df["Transaction Type"] == "Cash") | (df["Transaction Type"] == "Installment/Leasing")]
print("number of ads:")
print(cash_inst["Transaction Type"].value_counts())
print("median Price_lakh:")
print(cash_inst.groupby("Transaction Type")["Price_lakh"].median())
print("median Age:")
print(cash_inst.groupby("Transaction Type")["Age"].median())                        
```

**T4d answer:** The 5 cities with the most ads are shown above. `Unknown` was left out because it is the placeholder from Task 3a and not a real city. Karachi is far ahead of every other city, followed by Lahore, Islamabad and Rawalpindi.

**T4f answer:** The Corolla GLI median price rises steadily with year, from 12.25 lakh (2010) to 23.2 lakh (2018). Over the 8 years between them, a car loses about (23.2 − 12.25) / 8 ≈ **1.4 lakh rupees for each year of age**.

**T4g answer:** Yes, some look wrong. The 2015 Prius at 357.5 lakh (3.6 crore) and the 2007 Surf at 410 lakh with only 3,222 km are far too high for those models, so they are probably typing errors with extra digits. The Land Cruisers (420 to 670 lakh) are very expensive but possible, so they should be checked, not deleted automatically.

**T4h answer:** Cash ads are the vast majority (19,634 vs 850) and have a higher median price (6.9 vs 3.0 lakh). Installment/Leasing cars are much newer (median age about 4 vs 13 years) but still cheaper, so age does not explain the lower price. A possible reason is that some installment ads show a down payment or monthly amount instead of the full price, so these prices should be treated with care.
