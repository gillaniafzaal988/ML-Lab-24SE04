### Task 06
What is the main difference between a tuple and a list?

**Answer:** 
Lists are **mutable**, meaning their elements can be modified, added, or removed after creation, and they use square brackets `[]`. Tuples are **immutable**, meaning their contents cannot be changed once created, and they use parentheses `()`.


### Task 07
Function: grab the domain from "user@domain.com".

```python
def grab_domain(email):
  # Split the email string at '@' and return the final element (the domain)
  domain_name = email.split('@')[-1]
  return domain_name


print("Q7:", grab_domain("user@domain.com"))
```

### Task 08
Function: return True if the word "dog" is in a string (case-insensitive).

```python
def find_dog(st):
  # Convert the string to lowercase to ensure a case-insensitive search for 'dog'
  is_dog_present = 'dog' in st.lower()
  return is_dog_present


print("Q8:", find_dog("Is there a DoG here?"))
```

### Task 09
Function: count how many times "dog" occurs in a string (case-insensitive).

```python
def count_dog(st):
  # Normalize string case and count how many times 'dog' appears
  dog_frequency = st.lower().count('dog')
  return dog_frequency


print(
    "Q9:",
    count_dog("This dog runs faster than that other dog, Dog!"),
)
```

### Task 10
caught_speeding(speed, is_birthday): <= 60 "No Ticket", 61-80 "Small Ticket", >= 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed, is_birthday):
  if is_birthday:
    adjusted_speed = speed - 5
  else:
    adjusted_speed = speed

  # Determine ticket severity based on the adjusted speed thresholds
  if adjusted_speed <= 60:
    ticket_status = "No Ticket"
  elif adjusted_speed <= 80:
    ticket_status = "Small Ticket"
  else:
    ticket_status = "Big Ticket"

  return ticket_status


print("Q10 (Birthday 81):", caught_speeding(81, True))
print("Q10 Output (Regular 81):", caught_speeding(81, False))
```
